### 📄 `modelos.py`

Modelo ORM que describe la tabla.

````python

%%writefile modelos.py

from sqlalchemy import Integer, String
from sqlalchemy.orm import DeclarativeBase, Mapped, mapped_column


class Base(DeclarativeBase):
    pass


class Libro(Base):
    __tablename__ = "libros"

    id: Mapped[int] = mapped_column(
        Integer,
        primary_key=True,
        autoincrement=True
    )

    titulo: Mapped[str] = mapped_column(
        String(80),
        nullable=False
    )

    anio: Mapped[int] = mapped_column(
        Integer,
        nullable=False
    )
````

### 📄 `alembic.ini`

Configuración de Alembic.

````python
%%writefile alembic.ini
[alembic]
script_location = migrations
sqlalchemy.url = sqlite:///biblioteca.db
````

### 📄 `migrations/env.py`

Puente entre Alembic y los modelos.

````python
%%writefile migrations/env.py
from logging.config import fileConfig
from alembic import context
from sqlalchemy import engine_from_config, pool
from modelos import Base

config = context.config
target_metadata = Base.metadata

# Completa la configuración de conexión y ejecución.
````

### 📄 `migrations/versions/0001_crear_libros.py`

Migración versionada.

````python

%%writefile migrations/versions/0001_crear_libros.py

from alembic import op
import sqlalchemy as sa

revision = "0001"
down_revision = None
branch_labels = None
depends_on = None


def upgrade():
    op.create_table(
        "libros",
        sa.Column("id", sa.Integer(), primary_key=True, autoincrement=True),
        sa.Column("titulo", sa.String(length=80), nullable=False),
        sa.Column("anio", sa.Integer(), nullable=False)
    )


def downgrade():
    op.drop_table("libros")
````

**Requisitos del contrato**

- El modelo Autor tiene id y nombre.
- El modelo Libro tiene id, titulo, anio y una llave foránea autor_id.

- GET /api/autores/{id} devuelve el autor con sus libros anidados.

> **Pista conceptual:** En SQLAlchemy, `ForeignKey('autores.id')` y `relationship('Libro', back_populates='autor')` describen la relación. La base se crea con `Base.metadata.create_all(engine)`.

### 📄 `app.py`

Archivo principal de la implementación.

````python

%%writefile app.py

from fastapi import FastAPI, HTTPException
from pydantic import BaseModel, Field
from sqlalchemy import create_engine, Column, Integer, String, ForeignKey
from sqlalchemy.orm import declarative_base, relationship, sessionmaker

app = FastAPI(title="Tu relación autor → libros se verá así")

# --------------------------------
# 1. Configuración de la base de datos
# --------------------------------

DATABASE_URL = "sqlite:///./biblioteca.db"

engine = create_engine(
    DATABASE_URL,
    connect_args={"check_same_thread": False}
)

SessionLocal = sessionmaker(
    autocommit=False,
    autoflush=False,
    bind=engine
)

Base = declarative_base()


# --------------------------------
# 2. Modelo ORM: Autor
# --------------------------------

class Autor(Base):
    __tablename__ = "autores"

    id = Column(Integer, primary_key=True, index=True)
    nombre = Column(String(150), nullable=False)

    libros = relationship(
        "Libro",
        back_populates="autor",
        cascade="all, delete-orphan"
    )


# --------------------------------
# 3. Modelo ORM: Libro
# --------------------------------

class Libro(Base):
    __tablename__ = "libros"

    id = Column(Integer, primary_key=True, index=True)
    titulo = Column(String(200), nullable=False)
    anio = Column(Integer, nullable=False)

    autor_id = Column(
        Integer,
        ForeignKey("autores.id"),
        nullable=False
    )

    autor = relationship(
        "Autor",
        back_populates="libros"
    )


# --------------------------------
# 4. Crear las tablas
# --------------------------------

Base.metadata.create_all(bind=engine)


# --------------------------------
# 5. Esquemas de entrada
# --------------------------------

class AutorEntrada(BaseModel):
    nombre: str = Field(min_length=1, max_length=150)


class LibroEntrada(BaseModel):
    titulo: str = Field(min_length=1, max_length=200)
    anio: int = Field(ge=0)


# --------------------------------
# 6. Registrar un autor
# --------------------------------

@app.post("/api/autores", status_code=201)
def crear_autor(datos: AutorEntrada):

    db = SessionLocal()

    try:
        autor = Autor(nombre=datos.nombre)

        db.add(autor)
        db.commit()
        db.refresh(autor)

        return {
            "id": autor.id,
            "nombre": autor.nombre
        }

    except Exception:
        db.rollback()
        raise HTTPException(
            status_code=500,
            detail="Error al registrar el autor"
        )

    finally:
        db.close()


# --------------------------------
# 7. Registrar un libro asociado
# --------------------------------

@app.post("/api/autores/{autor_id}/libros", status_code=201)
def crear_libro(autor_id: int, datos: LibroEntrada):

    db = SessionLocal()

    try:
        autor = db.query(Autor).filter(
            Autor.id == autor_id
        ).first()

        if autor is None:
            raise HTTPException(
                status_code=404,
                detail="Autor no encontrado"
            )

        libro = Libro(
            titulo=datos.titulo,
            anio=datos.anio,
            autor=autor
        )

        db.add(libro)
        db.commit()
        db.refresh(libro)

        return {
            "id": libro.id,
            "titulo": libro.titulo,
            "anio": libro.anio,
            "autor_id": libro.autor_id
        }

    except HTTPException:
        raise

    except Exception:
        db.rollback()
        raise HTTPException(
            status_code=500,
            detail="Error al registrar el libro"
        )

    finally:
        db.close()


# --------------------------------
# 8. Consultar autor y sus libros
# --------------------------------

@app.get("/api/autores/{autor_id}")
def obtener_autor(autor_id: int):

    db = SessionLocal()

    try:
        autor = db.query(Autor).filter(
            Autor.id == autor_id
        ).first()

        if autor is None:
            raise HTTPException(
                status_code=404,
                detail="Autor no encontrado"
            )

        return {
            "id": autor.id,
            "nombre": autor.nombre,
            "libros": [
                {
                    "id": libro.id,
                    "titulo": libro.titulo
                }
                for libro in autor.libros
            ]
        }

    finally:
        db.close()


# --------------------------------
# 9. Consultar todos los libros
# --------------------------------

@app.get("/api/libros")
def listar_libros():

    db = SessionLocal()

    try:
        libros = db.query(Libro).all()

        return {
            "libros": [
                {
                    "id": libro.id,
                    "titulo": libro.titulo,
                    "anio": libro.anio,
                    "autor_id": libro.autor_id
                }
                for libro in libros
            ]
        }

    finally:
        db.close()
````

### 📄 `app.py`

Archivo principal de la implementación.

````python

%%writefile app.py
from fastapi import FastAPI, HTTPException, Response, Cookie
from pydantic import BaseModel
import secrets

app = FastAPI(title="Manejo de Sesiones Stateful")

# Usuarios registrados para realizar las pruebas
usuarios = {
    "ana": "secreta"
}

# Diccionario para guardar las sesiones activas
sesiones = {}

# Modelo para recibir los datos del login
class Login(BaseModel):
    usuario: str
    password: str


# Ruta para iniciar sesión
@app.post("/api/login")
def iniciar_sesion(datos: Login, response: Response):

    if usuarios.get(datos.usuario) != datos.password:
        raise HTTPException(
            status_code=401,
            detail="Usuario o contraseña incorrectos"
        )

    # Crear un identificador único para la sesión
    session_id = secrets.token_urlsafe(32)

    # Guardar la sesión en el servidor
    sesiones[session_id] = datos.usuario

    # Enviar la cookie al cliente
    response.set_cookie(
        key="session_id",
        value=session_id,
        httponly=True,
        samesite="lax",
        path="/"
    )

    return {"mensaje": "sesión iniciada"}


# Ruta para consultar el perfil
@app.get("/api/perfil")
def obtener_perfil(session_id: str = Cookie(default=None)):

    if not session_id or session_id not in sesiones:
        raise HTTPException(
            status_code=401,
            detail="No hay una sesión válida"
        )

    return {"usuario": sesiones[session_id]}


# Ruta para cerrar sesión

@app.post("/api/logout", status_code=204)
def cerrar_sesion(
    response: Response,
    session_id: str = Cookie(default=None)
):

    # Eliminar la sesión del servidor
    if session_id:
        sesiones.pop(session_id, None)

    # Eliminar la cookie del navegador
    response.delete_cookie(
        key="session_id",
        httponly=True,
        samesite="lax",
        path="/"
    )

    # Establecer explícitamente el código HTTP 204
    response.status_code = 204

    return response
````
